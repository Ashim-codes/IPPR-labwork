## **Lists (Mutable , ordered)**

Basic List Operations



```
fruits = ["apples", "banana", "orange","mango"]
print("Original List :",fruits)
print("First Fruit :", fruits[0])
print("Last Fruit :", fruits[-1])


# Adding items
fruits.append("cherry")
print("Updated List :",fruits)


#Removing items
fruits.remove("banana")
print("After Removing:",fruits)


#slicing
print(fruits[2:3])
print(fruits[:3])
```


List with numbers




```
numbers =[1,2,4,5,6,23,65]
print("Sum:",sum(numbers))
print("Max among numbers:", max(numbers))
print("Min among numbers: ", min(numbers))
print("Sorted :",sorted(numbers))
print("Length:", len(numbers))
```

# **Basic Dicitionary Operations**

```
students = {
    "name": "Ashim",
    "age": 21,
    "course": "BCE",
    "grade" : "A"
}

print("Student Info:", students)
print("Name:",students["name"])
print("Age:",students["age"])
print("keys:",students.keys())
print("Values:",students.values())




```

```
#Adding or Updating

students["city"]="Kathmandu"
students["grade"] ="A+"
students.update({"car":"B.Y.D"})
print("Updated info :", students)
```

```
products={
    "apple":0.99,
    "samsung":1,
    "xiomi":2
}

products.update({"grape":1.99})
print("Products",products)


#Display all products and price
print("Products : price")
for prod , price in products.items():

  print(f"{prod} : ${price}")
```

# **Tuples (immutable,ordered)**

Create a tuple of days in a week

```
weeks = ("Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday")
print("Weekend Days:",weeks[5:7])
print("Week Days:",weeks[0:7])

print("Weekend Days:", weeks[-2:])
print("Week Days:", weeks[-7:-2])
print("Reverse",weeks[::-1])
```
